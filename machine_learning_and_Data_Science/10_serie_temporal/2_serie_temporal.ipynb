
# Série Temporal - Introdução

## O que é Série Temporal?

Uma série temporal é um conjunto de dados organizados ao longo do tempo.

Exemplos:
- Temperatura diária
- Preço de ações
- Quantidade de vendas mensais
- Número de passageiros por mês

## Objetivos
Neste notebook você aprenderá:

1. O que é uma série temporal
2. Como carregar dados
3. Como converter datas
4. Como visualizar gráficos
5. Tendência
6. Sazonalidade
7. Média móvel


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (12,6)
```


# Criando um dataset simples

```python
datas = pd.date_range(start = '2020-01-01', periods=24, freq='M')
vendas = [100,120,130,140,150,160,170,180,175,190,200,220,210,230,240,250,260,270,280,290,300,320,330,350]
base = pd.DataFrame({'data':datas,'vendas':vendas})

base.head()
```


# Transformando em Série Temporal

```python
base = base.set_index('data')

base.head()
```

# Visualizando os dados


```python
base.plot()
plt.title('Vendas ao longo do tempo')
plt.xlabel('Data')
plt.ylabel('Vendas')
```
