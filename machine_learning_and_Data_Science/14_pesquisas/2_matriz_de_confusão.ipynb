# A matriz de confusão (confusion matrix) é uma tabela usada em Machine Learning para avaliar modelos de classificação.

Ela mostra:

- quantas previsões o modelo acertou
- quantas errou
- e quais tipos de erro aconteceram

```python
from sklearn.metrics import confusion_matrix
```

```python
y_real = [1,0,1,1,0]
y_pred = [1,0,0,1,1]
```

```python
matriz = confusion_matrix(y_real, y_pred)
print(matriz)
```

Interpretação: [[VN FP]
 [FN VP]]

Quando usar?

# A matriz de confusão é essencial em:

- classificação binária
- classificação multiclasse
- avaliação de modelos

Muito usada em:

diagnósticos médicos
detecção de fraude
spam
reconhecimento facial
análise de crédito
Em resumo

A matriz de confusão:

- ✅ mostra acertos e erros do modelo
- ✅ ajuda a entender os tipos de erro
- ✅ gera métricas importantes
- ✅ é fundamental em classificação
