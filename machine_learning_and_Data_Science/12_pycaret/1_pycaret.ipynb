
# Guia Completo de PyCaret com 5 Exemplos

## O que é PyCaret?

PyCaret é uma biblioteca low-code de Machine Learning para Python.

Ela automatiza:
- treinamento;
- comparação de modelos;
- preprocessamento;
- previsão;
- análise de dados.

Com poucas linhas de código é possível criar modelos completos.
