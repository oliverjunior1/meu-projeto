## O que é PyCaret?

PyCaret é uma biblioteca low-code de Machine Learning para Python.

Ela automatiza:
- treinamento;
- comparação de modelos;
- preprocessamento;
- previsão;
- análise de dados.

Com poucas linhas de código é possível criar modelos completos.

````python
from pycaret.classification import *
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

base = pd.DataFrame(iris.data, columns=iris.feature_names)
base['target'] = iris.target

base.head()
````

````python
clf = setup(data=base, target='target', session_id=123)
````

````python
melhor_modelo = compare_models()
````

````python
modelo_rf = create_model('rf')
````

````python
predicoes = predict_model(modelo_rf)
predicoes.head()
````


# Exemplo 3 - Clustering

## Objetivo
Agrupar dados semelhantes.

````python
from pycaret.clustering import *
from sklearn.datasets import load_wine

wine = load_wine(as_frame=True)

base_cluster = wine.frame

base_cluster.head()
````

````python
cluster = setup(
    data = base_cluster,
    session_id=123
)
````

````python
kmeans = create_model('kmeans')
````

````python
resultado_cluster = assign_model(kmeans)
resultado_cluster.head()
````


# Exemplo 5 - Série Temporal

## Objetivo
Prever valores futuros.


````python
from pycaret.time_series import *
import pandas as pd

datas = pd.date_range(start='2020-01-01', periods=36, freq='M')

vendas = [
    100,110,120,130,140,150,
    160,170,180,190,200,210,
    220,230,240,250,260,270,
    280,290,300,310,320,330,
    340,350,360,370,380,390,
    400,410,420,430,440,450
]

serie = pd.Series(vendas, index=datas)

serie.head()
````

````python
ts = setup(
    data=serie,
    fh=6,
    session_id=123
)
````

````python
melhor_ts = compare_models()
````

````python
modelo_ts = create_model('arima')
````

````python
forecast = predict_model(modelo_ts)
forecast
````
